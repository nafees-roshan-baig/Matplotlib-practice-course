```python
import matplotlib.pyplot as plt
a = [1,20,3,40,5]
b = [10,20,30,40,50]
plt.plot(a,b)
```

```python
import numpy as np
x = np.array([1,8])
y = np.array([3,10])
plt.plot(x,y)
plt.show()
```

```python
# plotting without a line(Markers only)
import numpy as np
x = np.array([1,8])
y = np.array([3,10])
plt.plot(x,y,'*')
```

```python
x = np.array([1,2,6,8])
y = np.array([3,8,1,10])

plt.plot(x,y,marker = '*')
plt.show
```

```python

y = np. array([3, 8, 1, 10])
plt.plot(y, marker = 'o')
```

```python
y = [3,8,1,10]
plt.plot(y,'o:m')
plt.show()
```

```python
#marker size (ms)
y = [3,8,1,10]
plt.plot(y,'*:c',ms =20)
plt.show()
```

```python
# Marker Edge Color (mec)
y = [3,8,1,10]
plt.plot(y,'*:c',ms =20,mec='m')
plt.show()
```

```python
y = [3,8,1,10]
plt.plot(y,'*:c',ms =20,mec='y',mfc='r')
plt.show()
```

```python
y = [3,8,1,10]
plt.plot(y,marker='o',ms=10,mec='r',mfc='g',c='k')
plt.show()
```

```python
# using multiple colours 
y = [3,8,1,10]
plt.plot(y,marker='v',ms = 10,mec='k',mfc='#de2f86',c = 'g')
plt.show()
```

```python
plt.plot(y,marker='o',ms=20,mec='blue',mfc='red')
plt.show()
```

```python
#line style
plt.plot(y,linestyle='dotted')
plt.show()
```
